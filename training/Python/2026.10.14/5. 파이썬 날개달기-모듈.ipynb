* 예외처리 trey-except 발생 오류와 변수 포함

```python
try:
    4 / 0
except ZeroDivisionError as e:
    print(e)
```

* 예외처리 try-finally 문

```python
try:
    f = open('foo.txt', 'w')
finally:
    f.close()
```

```python
try:
    print("나누기전")
    4 / 0
    print("나누기 후")
except ZeroDivisionError:
    print("오류 발생")
finally:
    print("finally 수행!!!")
```

```python
try:
    a = [1, 2]
    print(a[3])
    4 / 0
except ZeroDivisionError:
    print("0으로 나눌 수 없어 오류 발생")
except IndexError:
    print("인덱싱 할 수 없습니다.")
finally:
    print("finally 수행!!!")
```

```python
try:
    a = [1, 2]
    print(a[3])
    4 / 0
except ZeroDivisionError as e:
    print(e)
except IndexError as e:
    print(e)
finally:
    print("finally 수행!!!")
```

```python
try:
    a = [1, 2]
    print(a[3])
    4 / 0
except (ZeroDivisionError, IndexError) as e:
    print(e)
```

```python
# try-else
try:
    age = int(input('나이를 입력하세요:'))
except:
    print('입력이 정확하지 않습니다.')
else:
    if age <= 18:
        print('미성년자입니다. 출입금지!!')
    else:
        print('환영합니다.')
```
