```python
#scientists.csv 로드하기
import pandas as pd
scientists = pd.read_csv('../data/scientists.csv')
scientists
```

```python
print(scientists['Age'].mean())
```

```python
scientists[scientists['Age'] > scientists['Age'].mean()]
```

```python
# 날짜가 문자열로 저장됨
print(scientists['Born'].dtype)
print(scientists['Died'].dtype)
```

```python
#시간관련 작업을 할 수 있도록 datetime 자료향으로 변형
born_datetime = pd.to_datetime(scientists['Born'], format = '%Y-%m-%d')
born_datetime
```

```python
died_time = pd.to_datetime(scientists['Died'], format = '%Y-%m-%d')
died_time
```

```python
scientists.head(n=2)
```

```python
scientists['born-dt'], scientists['died-dt'] = (born_datetime, died_time)
scientists
```

```python
scientists.shape
```

```python
scientists['age_days_dt'] = (scientists['died-dt'] - scientists['born-dt'])
scientists
```
