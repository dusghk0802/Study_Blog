## get 방식으로 요청할 때 파라미터를 전달하는 방법

```python
# parameters = {"code": "005930"} # 삼성전자
parameters = {"code": "000270"} # 기아차
res = requests.get("https://finance.naver.com/item/main.naver", params = parameters)
print(res.url)
```

## post 방식으로 요청

```python
res = requests.post("http://finance.naver.com/item/main.naver?code=00270")
print(res.url)
```

## HTML 소스 데이터로 변환기

```python
#!pip install bs4
```

```python
from bs4 import BeautifulSoup
html_txt = "<p class = ‘wether’ id = ‘tw’>오늘의 날씨</p> <h1> 한때 소나기가 내리겠습니다. </h1>"
soup = BeautifulSoup(html_txt, "html.parser") # html 파서로 파싱하기, 분석 가능한 데이터 형태로 변환
soup
```

```python
tag = soup.find("p")
print(tag)
print(tag.name)
print(tag.attrs)
print(tag.attrs["class"]) # tag 속성 중 class만 출력
print(tag.attrs["id"]) # tag 속성 중 id만 출력
print(tag.text)
```

```python
from bs4 import BeautifulSoup  # BeautifulSoup 모듈 불러오기

# html 소스 생성
html_txt = """
<html>
<head><title>BS page</title></head>
<body>

<h1 class="portal_cls">검색 포털1</h1>
<h1 class="portal_cls">검색 포털2</h1>
<h1 class="portal_cls">검색 포털3</h1>
<p>
<a href="http://www.daum.net">다음 바로가기</a><br>
<a href="http://www.naver.com">네이버 바로가기</a>
</p>
<a href="http://www.google.com" class="alink_cls">구글</a>
<a target="_테스트_">테스트</a>
<p class="footage_cls" id="company">2021, ABC Company</p>
<p class="footage_cls" id="addr">Korea</p>
</body>
</html>
"""
soup = BeautifulSoup(html_txt, "html.parser")
```

```python
tag = soup.select_one("h1")
print(tag.text)
```

```python
tag_list = soup.select("h1")
for tag in tag_list:
    print(tag.text)
```

```python
tag_list = soup.select("body p > a") #<body> <p> 태그 하위 <a> 태그 추출
for tag in tag_list:
    print(tag.text)
```

```python
tag_list = soup.select(".footage_cls") # .footage_cls class 추출
for tag in tag_list:
    print(tag.text)
```

```python
print(tag_list[1].text)
```

```python
tag_list = soup.select("#addr") #addr ID로 추출
for tag in tag_list:
    print(tag.text)
```

```python
tag_list = soup.select("a[href]") #a[href]추출
for tag in tag_list:
    print(tag.text, tag.attrs["href"])
```

```python
from selenium import webdriver  # webdriver 모듈 불러오기
from selenium.webdriver.common.by import By # selenium의 By 클래스 불러오기

URL = "https://www.naver.com"  # 네이버 URL을 URL 변수에 저장
driver = webdriver.Chrome()  # driver 객체 생성
driver.get(URL)  # URL에 저장된 주소를 크롬에서 열기

input_tag = driver.find_element(By.CSS_SELECTOR, "#query")  # 검색창으로 이동
input_tag.send_keys("방탄소년단")  # 검색창에 문자 입력
input_tag.send_keys("\n")  # Enter 키 입력
driver.implicitly_wait(3)  # 최대 3초 대기
```
