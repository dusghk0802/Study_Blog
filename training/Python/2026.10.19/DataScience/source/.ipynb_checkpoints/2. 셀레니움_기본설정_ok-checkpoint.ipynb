## 1. 필요한 라이브러리 가져오기


```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
```

## 2. Chrome 옵션 설정

- `detach=True` : Python 코드 실행이 끝나도 브라우저를 유지합니다.
- `excludeSwitches` : 불필요한 Chrome 로그 메시지를 줄입니다.


```python
chrome_options = Options()

# 코드 실행이 끝나도 브라우저 유지
chrome_options.add_experimental_option("detach", True)

# 불필요한 Chrome 로그 메시지 제거
chrome_options.add_experimental_option(
    "excludeSwitches", ["enable-logging"]
)
```

## 3. ChromeDriver 설정 및 브라우저 실행

`webdriver-manager`를 이용해 ChromeDriver를 자동으로 설치하고 실행합니다.


```python
service = Service(ChromeDriverManager().install())

driver = webdriver.Chrome(
    service=service,
    options=chrome_options
)
```

## 4. 웹 페이지 접속


```python
driver.get("https://www.naver.com")
```

## 5. 브라우저 종료

- `driver.close()` : 현재 브라우저 창 하나를 닫습니다.
- `driver.quit()` : Selenium이 실행한 전체 브라우저와 WebDriver 세션을 종료합니다.



```python
import requests				# requests 모듈 불러오기
res = requests.get("http://www.naver.com")	                 # get으로 네이버 홈페이지 가져오기
#print(res.text)  
print("\n".join(res.text.splitlines()[:10]))
```

```python
parameters = {"code" : "005930"}
res = requests.get("https://finance.naver.com/item/sise.nhn", params = parameters)
print (res.url)
#print (res.text)
print("\n".join(res.text.splitlines()[:10]))
```

```python
# 실습이 끝난 후 실행하세요.
driver.quit()
#driver.close()
```
