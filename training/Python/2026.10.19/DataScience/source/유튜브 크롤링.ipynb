## Selenium에서 Bequtifulsoup 사용 이유, 사용방법
## 데이터 전처리(preprocessing) --> 조회수를 숫자로 바꾸기

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time

# 브라우저 꺼짐 방지
chrome_options = Options()
chrome_options.add_experimental_option("detach", True)

# 크롬 드라이버 생성
driver = webdriver.Chrome(options=chrome_options)
```

```python
def text_to_num(text):
    text = text.replace("조회수", "").replace("회","").strip()
    if "억" in text:
        num = float(text.replace("만", "")) * 100000000
    elif "만" in text:
        num = float(text.replace("만", "")) * 10000
    elif "천" in text:
        num = float(text.replace("천", "")) * 1000
    elif "없음" == text:
        num = 0
    else:
        num = int(text)
    return num
```

```python
keyword = "울릉도" 
```

```python
url =f"https://www.youtube.com/results?search_query={keyword}"
driver.get(url)
```

```python
scroll_count = 2
```

```python
i = 1
while True:
    # 맨 아래로 스크롤을 내린다. 끝까지 스크롤 (END)
    driver.find_element(By.CSS_SELECTOR, "body").send_keys(Keys.END)
    #스크롤 사이에 페이지 로딩 시간
    time.sleep(2)
    
    if i == scroll_count:
        break
    i += 1
```

```python
#Selenium - BeautifulSoup 연동방법
html = driver.page_source
soup = BeautifulSoup(html, 'html.parser')
infos = soup.select("div.text-wrapper")

result = []
for i, info in enumerate(infos, 1):  #1 start
    # 원하는 정보 가져오기
    # 제목
    title = info.select_one("a#video-title").text
    
    try:
        #조회수
        views = info.select_one("div#metadata-line > span:nth-of-type(1)").text
        #날짜
        date = info.select_one("div#metadata-line > span:nth-of-type(2)").text
    except:
        views = '조회수 0회'
        date = '날짜 없음'
    
    views = text_to_num(views)  ###
    print(title, views, date)
    result.append([i, title, views, date])
   
```

```python
import pandas as pd
results_df = pd.DataFrame(result)
results_df.columns = ['번호', '제목', '조회수', '날짜']
results_df.to_excel('./data/youtubecrawing.xlsx')
```

* nth-child는 몇번째 자식인지로 선택

* nth-of-type은 해당 태그들 중에 몇번째인지로 선택

## 데이터 전처리
* 조회수 15만회 
* replace("조회수", "")
* 15만회 
* replace("회", "") + strip()
* 15 만 
* 숫자 자료형으로 변환 후 10000 곱하기
