## 2-5 While 문

```python
treeHit = 0
while treeHit < 10:
    treeHit = treeHit + 1
    print(f"나무를 {treeHit}번 찍었습니다.")
    if treeHit == 10:
        print("나무 넘어갑니다.")
```

```python
# 커피 자판기
coffee = 10
money = 300
while money:
    print("돈을 받았으니 커피를 줍니다.")
    coffee = coffee - 1
    print (f"남은 커피의 양은 {coffee}개 입니다.")
    if coffee == 0:
        print("커피가 떨어짐. 판매를 중지합니다.")
        break
```

```python
# coffee.py
coffee = 2
while True:
    money = int(input("돈을 넣어 주세요"))
    if money == 300:
        print("커피를 줍니다.")
        coffee = coffee - 1
    elif money > 300:
        print(f"거스름돈 {money - 300}를 주고 커피를 줍니다.")
        coffee = coffee - 1
    else:
        print("돈을 다시 돌려주고 커피를 주니 않습니다.")
        print(f"남은 커피의 양은 {coffee}개 입니다.")
    if coffee == 0:
        print("커피가 다 떨어졌습니다. 판매 중지합니다.")
        break
```

```python
#1~10까지 숫자 중에서 홀수만 출력하는 While문 작성
a = 0
while a < 10:
    a = a + 1
    if a % 2 == 0: continue
    print(a)
```

```python
marks = [90, 25, 45, 80]
number = 0
for mark in marks:
    number = number + 1
    if mark >= 60:
        print(f"{number}번 학생은 합격입니다.")
    else:
        print(f"{number}번 학생은 불합격입니다.")
```

```python
i, hap = 0, 0
num1, num2, num3 = 0, 0, 0

num1 = int(input("시작값을 입력하세요: "))
num2 = int(input("끝값을 입력하세요: "))
num3 = int(input("증가값을 입력하세요: "))

for i in range(num1, num2 + 1, num3):
    hap = hap + i
print(f"{num1}에서 {num2}까지 {num3}씩 증가시킨 값의 합계: {hap}")
```

```python
marks = [90, 25, 67, 45, 80]
for number in range(len(marks)): #0~5미만
    if marks[number] < 60:
        continue
    print(f"{number+1}번 학생 축하합니다. 합격입니다.")
```

</br>1~100의 합계를 구하되 3의 배수는 제외하고 더하기

```python
sum, i = 0, 0
for i in range(1, 101):
    if i % 3 ==0:
        continue
    sum += i
print(f"0과 100사이의 3의 배수제외한 합계: {sum}")
```
