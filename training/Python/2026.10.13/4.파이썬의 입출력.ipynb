```python
def add_many(*args):
    result = 0
    for i in args:
        result = result + i
    return result
```

```python
result = add_many(1, 2, 3) #입력값-3개
result
```

```python
result = add_many(1, 2, 3, 4, 5, 6, 7, 8, 9) #입력값-9개
result
```

```python
def add_mul(choice, *args):
    if choice == "add":
        result = 0
        for i in args:
            result = result + i
    elif choice == "mul":
        result = 1
        for i in args:
            result = result * i
    return result
```

```python
result = add_mul('add', 1, 2, 3, 4, 5)
result
```

```python
result = add_mul('mul', 1, 2, 3, 4, 5)
result
```

```python
def say_myself(name, age, man=True):
    print(f"나의 이름은 {name}입니다.")
    print(f"나이는 {age}살 입니다.")
    if man:
        print("남자입니다.")
    else:
        print("여자입니다.")
```

```python
say_myself("김코사", 23)
```

```python
say_myself("김코사", 23, True)
```

```python
say_myself("김코사", 23, False)
```

```python
# 매개변수 초깃값을 설정할 때 주의 - 맨마지막에 위치하도록
def say_myself(name, man=True, age): # 오류 발생
    print(f"나의 이름은 {name}입니다.")
    print(f"나이는 {age}살 입니다.")
    if man:
        print("남자입니다.")
    else:
        print("여자입니다.")
```

```python
# 함수의 스코프
a = 1
def vartest(a):
    a = a + 1
```

```python
vartest(a)
print(a)
```

```python
#1. return문 사용
a = 1
def vartest(a):
    a = a + 1
    return a
```

```python
a = vartest(a)
print(a)
```

```python
#2. global 명령어 사용
a = 1
def vartest():
    global a
    a = a + 1
```

```python
vartest()
print(a)
```

```python
add = lambda a, b: a + b
result = add(3, 4)
print(result)
```

```python
def add(a, b):
    return a + b
result = add(3, 4)
print(result)
```

```python
f = open("c:\doit\새파일.txt", 'r')
while True:
    line = f.readline()
    if not line: break
    print(line)
f.close()
```

```python
f = open("c:\doit\새파일.txt", 'r')
lines = f.readlines()
for line in lines:
    line = line.strip() # 줄 바꿈 제거
    print(line)
f.close()
```

```python
with open("c:\\doit\\dream1.txt", 'r') as my_file:
    contents = my_file.read()
    word_list = contents.split(" ") # 빈칸 기준으로 단어를 분리 리스트
    line_list = contents.split("\n") # 한줄씩 문단을 분리하여 리스트

print("총 글자의 수:", len(contents))
print("총 단어의 수:", len(word_list))
print("총 줄의 수:", len(line_list))
```
