```python
# 함수를 이용해서 계산기의 더하기 기능을 구현
result = 0

def add(num):
    global result
    result += num
    return result
```

```python
print(add(3))
print(add(4))
```

```python
# 만갹, 2대의 계산기가 필요하다면
result1 = 0
result2 = 0

def add1(num):
    global result1
    result1 += num
    return result1

def add2(num):
    global result2
    result2 += num
    return result2
```

```python
print(add1(3))
print(add1(4))
print(add2(8))
print(add2(10))
```

```python
# 클래스 만들기-Calculator
class Calculator:
    def __init__(self):
        self.result = 0

    def add(self, num):
        self.result += num
        return self.result

    def sub(self, num):
        self.result -= num
        return self.result
```

```python
cal1 = Calculator()
cal2 = Calculator()
cal3 = Calculator()
```

```python
print(cal1.add(3))
print(cal1.add(4))
print(cal2.add(10))
print(cal2.add(8))
print(cal3.add(2))
print(cal3.add(66))
```

```python
# mul, sub, div 메소드 추가
class FourCal:
    def setdata(self, first, second):
        self.first = first
        self.second = second
        
    def add(self):
        result = self.first + self.second
        return result

    def mul(self):
        result = self.first * self.second
        return result

    def sub(self):
        result = self.first - self.second
        return result

    def div(self):
        result = self.first / self.second
        return result
```

```python
a = FourCal()
b = FourCal()
a.setdata(4, 2)
b.setdata(3, 8)
```

```python
print(a.add())
print(a.mul())
print(a.sub())
print(a.div())
```

```python
print(b.add())
print(b.mul())
print(b.sub())
print(b.div())
```

```python
class FourCal:
    def __init__(self, first, second):
        self.first = first
        self.second = second

    def setdata(self, first, second):
        self.first = first
        self.second = second
        
    def add(self):
        result = self.first + self.second
        return result

    def mul(self):
        result = self.first * self.second
        return result

    def sub(self):
        result = self.first - self.second
        return result

    def div(self):
        result = self.first / self.second
        return result
```

```python
a = FourCal(4, 2)
print(a.first)
```

```python
a = FourCal(4, 2)
a.add()
```

```python
a.div()
```

```python
class MoreFourcal(FourCal):
    pass
```

```python
a = MoreFourcal(4, 2)
print(a.add())
print(a.mul())
print(a.sub())
print(a.div())
```

```python
class MoreFourcal(FourCal):
    def pow(self):
        result = self.first ** self.second
        return result
```

```python
a = MoreFourcal(4, 2)
print(a.pow())
print(a.add())
```

* 메서드 오버라이딩

```python
# a = FourCal(4, 0) ==> 4/0이므로 오류 발생
# a.div()
```

```python
class SafeFourCal(FourCal):
    def div(self):
        if self.second == 0:
            return 0
        else:
            return self.first / self.second
```

```python
a = SafeFourCal(4, 0)
a.div()
```
