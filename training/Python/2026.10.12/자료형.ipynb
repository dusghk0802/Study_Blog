## 2-6 집합 자료형

```python
s1 = set([1, 2, 3])
print(s1)
print(type(s1))
```

```python
li = list(s1)
print(li)
print(type(li))
li[0]
```

```python
#집합 자료형은 중복을 허용하지 않습니다.
s2 = set('Hello')
s2
```

```python
s1 = set([1, 2, 3, 4, 5, 6])
s2 = set([4, 5, 6, 7, 8, 9])
```

```python
#교집합 구하기
print(s1 & s2)
print(s1.intersection(s2))
```

```python
# 합집합
print(s1 | s2)
print(s1.union(s2))
```

```python
# 차집합
print(s1 - s2)
print(s2 - s1)
print(s1.difference(s2))
print(s2.difference(s1))
```

```python
# 값 1개 추가 - add
s1 = set([1, 2, 3])
s1.add(4)
s1
```

```python
# 값 여러개 추가 - update
s1 = set([1, 2, 3])
s1.update([4, 5, 6])
s1
```

```python
# 특정값 제거하기 -remove
s1 = set([1, 2, 3])
s1.remove(2)
s1
```

```python
# 특정값 제거하기 -discard
s1 = set([1, 2, 3])
s1.discard(2)
print(s1)
s1.discard(4) #없는 값을 제거해도 오류가 발생하지 않음
print(s1)
```

```python
# 모든 값 제거하기 -clear
s1 = set([1, 2, 3])
s1.clear
s1
```

```python
# 리스트에서 중복 값 제거
a = [1, 1, 1, 2, 2, 3, 3, 3, 4, 4, 5]
aSet = set(a)
b = list(aSet)
print(b)
```

## 2-8 자료형의 값을 저장하는 공간, 변수
* 1) [:] 이용하기

```python
a = [1, 2, 3]
print(a)
print(id(a)) #a가 가르키는 메오리의 주소
```

```python
b = a
print(id(a))
print(id(b))
```

```python
a[1] = 4
print(a)
print(b) #a만 변경하고 싶었는데 b도 변경됨
```

```python
# [:] 이용하기
a = [1, 2, 3]
b = a[:]
a[1] = 4
print(a) #a[1]만 4로 변경됨
print(b) #a리스트 값을 변경하더라도 b리스트에는 영향이 없다. [:]
```

* copy 모듈 이용하기

```python
from copy import copy
a = [1, 2, 3]
b = copy(a) #2)
a[1] = 4
print(a)
print(b)
```
