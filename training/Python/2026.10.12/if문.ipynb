1. 100000만원을 초과한 금액에 대해 5%를 할인해 주고 있습니다.
</br>총 금액이 110000일 경우 아래와 같은 프로그램을 작성해 보세요.
</br>===========결과===============
</br>구입 금액을 입력하시오:110000
</br>지불 금액은 104500.0입니다.

```python
price = int(input("구입 금액을 입력하시오:"))
if price > 100000:
    price = price * 0.95
print(f"지불 금액은 {price}입니다.")
```

3.월을 입력받아 2월이면 "월의 날수는 28일"
</br>4,6,10월이면 월의 날수는 30일
</br>그외는 31일 이라고 출력하세요.

==============================
</br>월을 입력하시오: 2
</br>월의 날수는 28일

```python
month = int(input("월을 입력하시오:"))
if month == 2:
    print("월의 날수는 28일")
elif month in [4, 6, 10]:
    print("월의 날수는 30일")
else:
    print("월의 날수는 31일")
```
