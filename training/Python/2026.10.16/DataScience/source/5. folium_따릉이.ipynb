## 공공자전거 데이터 지도 시각화

```python
import folium
import pandas as pd
import json
import googlemaps
```

```python
#!pip install googlemaps
```

```python
cyc = pd.read_excel('../data/공공자전거 대여소 정보.xlsx')
cyc.head(1)
```

```python
from folium import plugins

minimap = plugins.MiniMap()

mapping = folium.Map(location=[cyc['위도'].mean(), cyc['경도'].mean()],
                     zoom_start=11)
mapping.add_child(minimap)

marker_cluster = plugins.MarkerCluster().add_to(mapping)

for n in cyc.index:
    name = "대여소명 : " + cyc.loc[n, '대여소명'] + " / 거치대 수 : " + str(cyc.loc[n, '거치대수'])
    folium.Marker([cyc['위도'][n], cyc['경도'][n]], tooltip=name).add_to(marker_cluster)

mapping
```
