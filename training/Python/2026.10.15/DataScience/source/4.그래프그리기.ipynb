## matplotlib
* tips 데이터셋

```python
# 손님이 식당에서 지불한 전체 금액과 팁, 성별, 흡연 여부, 요일, 시간, 식사 인원
tips = sns.load_dataset("tips")
print(tips)
print(type(tips))
```

### 일변량 그래프 그리기
* 통계 용어로 변수가 1개일 때 이를 '일변량'이라고 함
* 변수를 하나만 사용해서 일변량 그래프를 그림

```python
fig = plt.figure()
axes1 = fig.add_subplot(1, 1, 1)
axes1.hist(data=tips, x='total_bill', bins=10)
axes1.set_title('Histogram of total_bill')
axes1.set_xlabel('Total Bill')
axes1.set_ylabel('Frequency')
plt.show()
```

### 이변량 그래프 그리기
* 변수 2개를 이변량이라고 함

### 산점도 그래프 그리기
* 산점도(scatter)는 연속 변수 사이의 관계를 나타내는 그래프

```python
fig = plt.figure()
axes1 = fig.add_subplot(1, 1, 1)

axes1.scatter(tips['total_bill'], tips['tip'])

axes1.set_title('Scatterplot of total_bill vs tip')
axes1.set_xlabel('Total Bill')
axes1.set_ylabel('Tip')
plt.show()
```

### 박스 그래프 그리기
* 박스(box) 그래프는 이산 변수와 연속 변수 사이의 관계를 나타냄
* 이산형 변수 - Female, Male처럼 몀확하게 구분되는 값을 의미
* 연속형 변수 - Tip와 같이 명확하게 셀 수 없는 변위의 값을 의미

```python
boxplot = plt.figure()
axes1 = boxplot.add_subplot(1, 1, 1)

axes1.boxplot(
    x= [
        tips[tips['sex'] == 'Female']['tip'],
        tips[tips['sex'] == 'Male']['tip']],
    labels = ['Female', 'Male'])

axes1.set_title('Scatterplot of total_bill vs tip')
axes1.set_xlabel('sex')
axes1.set_ylabel('Tip')
plt.show()
```

## 변수가 여러개인 그래프 그리기

```python
colors = {"Female": "#f1a340", "Male": "#998ec3"}

scatter_plot = plt.figure()
axes1 = scatter_plot.add_subplot(1, 1, 1)

axes1.scatter(data = tips,
              x = 'total_bill',
              y = 'tip',
              s = tips['size']**2*10,
              c = tips['sex'].map(colors),
              alpha=0.5)

axes1.set_title('Colored by Sex and Sized by Size')
axes1.set_xlabel('Total Bill')
axes1.set_ylabel('Tip')

scatter_plot.suptitle('Total Bill vs Tip')
plt.show()
```

```python
#map()은 인수로 전달한 딕셔너리의 키와 일치하는 행 데이터를 딕셔너리의 값으로 바꾸는 역할을 함
colors = {"Femal": "#f1a340", "Male": "#998ec3"}
tips['sex'].map(colors)
```

```python
# 성별을 새 변수로 추가(문자열은 산점도 그래프의 색상을 지정하는 값을 사용할 수 없다.)
def recode_sex(sex):
    if sex == 'Female':
        return 0
    else:
        return 1
```

### seaborn으로 그래프 그리기

```python
import seaborn as sns
tips = sns.load_dataset("tips")
```

```python
sns.histplot(data=tips, x = 'total_bill')
plt.show()
```

```python
hist1, ax = plt.subplots() # 그림 영역과 하위 그래프 객체를 한번에 생성

# ax=ax ax에는 생성한 하위 그래프 객체를 지정
sns.histplot(data = tips, x = 'total_bill', ax =  ax)
ax.set_title('Total Bill Histogram')
plt.show()
```

```python
den1, ax = plt.subplots()
sns.kdeplot(data = tips, x = 'total_bill', ax = ax)

ax.set_title('Total Bill Density')
ax.set_xlabel('Total Bill')
ax.set_ylabel('Unit Probability')

plt.show()
```

## 러그 그래프 그리기
* 러그(rug) 그래프는 변수 분포를 1차원으로 나타냄 

```python
rug, ax = plt.subplots()
sns.rugplot(data = tips, x = 'total_bill', ax = ax)
sns.histplot(data = tips, x = 'total_bill', ax = ax)

ax.set_title('Rug plot and hiostogram of Total Bill')

plt.show()
```

### 분포 그래프 그리기
* sns.displot()함수를 사용하면 어려 개의 일변량 그래프를 하나의 그래프로 표현할 수 있습니다.
* 히스토그램, 밀도 분포 그래프, 러그 그래프를 한번에 그림
* True, False 지정 가능

```python
fig = sns.displot(data = tips, x = 'total_bill', kde = True, rug = False)
plt.show()
```

### 막대 그래프 그리기
* 막대그래프는 히스토그램과 비슷하지만 구간별로 나누어 분포를 표현하지 않고 작대로 이산변수로 개수를 표현한다는데 차이가 있습니다.
* seaborn에서는 막대그래프를 count 그래프라고 함.

```python
count, ax = plt.subplots()
sns.countplot(data = tips, x = 'day', palette = 'Set2', ax = ax) # viridis, Set2, "husl", 'Spectral'

ax.set_title('Count of days')
ax.set_xlabel('Day of the week')
ax.set_ylabel('Frequency')

plt.show()
```

```python
count, ax = plt.subplots()
sns.countplot(data = tips, y = 'day', palette = 'viridis', ax = ax) # viridis, Set2, "husl", 'Spectral'

ax.set_title('Count of days')
ax.set_xlabel('Day of the week')
ax.set_ylabel('Frequency')

plt.show()
```
